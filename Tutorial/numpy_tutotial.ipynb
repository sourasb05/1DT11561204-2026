# NumPy Tutorial: Basics for Machine Learning


**NumPy** (short for Numerical Python) is a powerful open-source Python library used for numerical computing. It provides support for working with arrays, matrices, and a variety of mathematical functions to perform operations on these data structures efficiently. NumPy is a fundamental package for scientific computing in Python and serves as the base for many other libraries like Pandas, SciPy, and scikit-learn.

### Key Features of NumPy:

1. N-dimensional Array (ndarray):** 

- Core data structure of NumPy.
- Provides efficient storage and operations for homogeneous data (all elements of the same type).

2. Mathematical Functions:

- Supports a wide range of mathematical operations, such as addition, subtraction, multiplication, division, trigonometry, and more.

3. Broadcasting:

- Enables operations on arrays of different shapes without requiring explicit loops.

4. Linear Algebra:

- Provides support for linear algebra operations such as matrix multiplication, eigenvalues, determinants, and solving linear equations.

5. Random Number Generation:

- Functions to generate random numbers, random samples, and perform random sampling.

6. Efficient Performance:

- Optimized for performance with implementations in C, allowing fast execution for large-scale numerical computations.

### Why Use NumPy?

- **Performance:** It is much faster than standard Python lists for numerical operations because of its optimized C implementation.

- **Memory Efficiency:** Consumes less memory compared to Python lists for storing large datasets.

- **Ease of Use:** Provides concise syntax for complex numerical operations.

- **Compatibility:** Integrates seamlessly with other scientific libraries like Pandas, Matplotlib, and SciPy.



18. Reshaping

The .reshape() function changes the structure of your data without changing the data itself. It’s like taking a piece of clay shaped like a rope and folding it into a rectangle; the amount of clay stays the same, but the form is different.

Operation: Changing the structural layout of the array (from $2 \times 3$ to $1 \times 6$) without modifying the underlying data.

```python

print(f"\n Original 2D Array:", array_2d)
reshaped_array = array_2d.reshape((1, 6))
print("\nReshaped Array:")
print(reshaped_array)
```

For a reshape to work, the number of elements must remain exactly the same. The mathematical product of the new dimensions must equal the total number of elements:$$\text{Total Elements} = \text{Rows} \times \text{Columns}$$In your case: $1 \times 6 = 6$. 

19. Flattening (Row-Major)

Operation: Collapsing a multi-dimensional array into a 1D array by reading row-by-row.


The .flatten() function is essentially the "undo" button for dimensions. It takes a multi-dimensional array (like a 2D matrix or a 3D cube) and collapses it into a single, simple one-dimensional list.

Regardless of how complex the structure was before, after flattening, everything sits in one long row.

```python
print(f"\n Original 2D Array:", array_2d)   
flattened = array_2d.flatten()
print("\nFlattened Array:", flattened)
```

When you call .flatten(), NumPy reads the array element by element and places them into a new 1D array. By default, it reads across the rows first.

If you want to read down the columns first (top-to-bottom, then left-to-right) instead of across the rows, you need to change the order parameter.

20. Flattening (Column-Major)

Operation: Collapsing the array into a 1D format using 'Fortran' order (reading down columns first).

```python
print(f"\n Original 2D Array:", array_2d)   
flattened_column_first = array_2d.flatten(order='F')
print(flattened_column_first)
```

### 8. Matrix Operations

28. Matrix Multiplication (Dot Product)

Operation: Performing a standard linear algebra matrix multiplication using np.dot.

```python
# Matrix multiplication using np.dot
matrix_a = np.array([[1, 2], [3, 4]])
matrix_b = np.array([[5, 6], [7, 8]])
matrix_product = np.dot(matrix_a, matrix_b)
print("\nMatrix Product (using np.dot):")
print(matrix_product)
```

29. Matrix Multiplication (@ Operator)

Operation: Performing matrix multiplication using the modern Python @ infix operator.

```python
# Matrix multiplication using @ operator
matrix_product_at = matrix_a @ matrix_b
print("\nMatrix Product (using @ operator):")
print(matrix_product_at)
```

30. Transposition

Operation: Swapping the rows and columns of a matrix (Transposing).

```python
# Transpose of a matrix
matrix_transpose = matrix_a.T
print("\nTranspose of Matrix A:")
print(matrix_transpose)
```

31. Matrix Inversion

Operation: Computing the multiplicative inverse of a square matrix using the linalg sub-module.

```python
# Inverse of a matrix
matrix_square = np.array([[1, 2], [3, 4]])
matrix_inverse = np.linalg.inv(matrix_square)
print("\nInverse of Matrix:")
print(matrix_inverse)
```

32. Determinant Calculation

Operation: Calculating the determinant of a square matrix.

```python
# Determinant of a matrix
matrix_determinant = np.linalg.det(matrix_square)
print("\nDeterminant of Matrix:", matrix_determinant)
```


# 10. Broadcasting


35. Broadcasting (Scalar)

Operation: Demonstrating broadcasting by adding a single number to an entire 2D matrix.

```python
# Add a scalar to a 2D array
broadcast_add = array_2d + 5
print("\nBroadcast Addition:")
print(broadcast_add)
```

36. Broadcasting (Array-on-Array)

Operation: Multiplying a 2D array by a smaller 1D array; NumPy "broadcasts" the smaller array to match the shape of the larger one.

```python
# Multiply a 1D array with a 2D array
print("\n Array 1D:", array_1d)
print(f"\n Original 2D Array:", array_2d)
broadcast_mult = array_2d * array_1d[:3]
print("\nBroadcast Multiplication:")
print(broadcast_mult)
```
